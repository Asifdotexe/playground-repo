```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('white')
plt.rcParams['figure.dpi'] = 200
```

```python
# setting a seed for reproducibility
np.random.seed(423)
```

```python
# Define group sizes to intentionally cause an imbalance
# For Simpson's paradox the imbalance is the "engine" that causes it to work
young_group_treatment_a = 50
old_group_treatment_a = 200
young_group_treatment_b = 200
old_group_treatment_b = 50
```

Here is the idea, within each group (Young and Old), treatment a did genuinely performed better. however, we give treatment A mostly to the "Old" group (i.e., group that has a natural lower success rate / harder to treat group)

Meanwhile, treatment B is mostly given to the "Young" group who is generally easier to treat i.e., higher success rates

When we later combine all the patients (ignoring age), the aggregate average for treatment A is dragged down by it's low-performing old patients, while B's overall average is lifted by it's high performing young patients

Result? even though A > B within every subgroup. B > A overall. That's the reveral in the Simpson's Paradox

but let's see it in action?

```python
# Define success probabilities
probability_of_success = {
    ('Young Group', 'Treatment A'): 0.9,
    ('Old Group', 'Treatment A'): 0.6,
    ('Young Group', 'Treatment B'): 0.8,
    ('Old Group', 'Treatment B'): 0.5
}
```

```python
def generate_group(age_group: str, treatment_group: str, sample_size: int) -> pd.DataFrame:
    """
    Generate binary outcomes based on the inputs

    :param age_group: String containing the age group category
    :param treatment_group: String containing the treatment group category
    :param sample_size: Int value refering to the sample size of that age and treatment group
    :returns: Pandas dataframe containing the dataframe of age_group, treatment_group and success binary variable
    """
    success_rate = probability_of_success[(age_group, treatment_group)]
    success = np.random.binomial(1, success_rate, sample_size)

    return pd.DataFrame({
        'age_group': age_group,
        'treatment_group': treatment_group,
        'success': success
    })
```

```python
df = pd.concat([
    generate_group('Young Group', 'Treatment A', young_group_treatment_a),
    generate_group('Old Group', 'Treatment A', old_group_treatment_a),
    generate_group('Young Group', 'Treatment B', young_group_treatment_b),
    generate_group('Old Group', 'Treatment B', old_group_treatment_b)
])
```

```python
df.sample(5)
```

```python
# Calculating the success rates for the age and treatment group
df.groupby(['age_group', 'treatment_group']) \
    .success.mean()\
    .unstack()\
    .round(3)
```

```python
# calculate the success rate for just the treatment groups
df.groupby('treatment_group').success.mean().round(3)
```

```python
sns.barplot(df, x='treatment_group', y='success', hue='age_group', errorbar=None)
plt.title("Success rate by treatment group and age group", fontweight='bold')
plt.ylim(0, 1)
plt.xlabel("Treatment group", fontweight='bold')
plt.ylabel("Success", fontweight='bold')
plt.legend(title='Age group')
plt.grid()
plt.show()
```

```python
sns.barplot(df, x='treatment_group', y='success', errorbar=None, color="lightgreen")
plt.title("Overall success rate (Aggregated)", fontweight='bold')
plt.ylim(0, 1)
plt.xlabel("Treatment group", fontweight='bold')
plt.ylabel("Success", fontweight='bold')
plt.grid()
plt.show()
```
